```python
df["transaction_count"].fillna(df["transaction_count"].median(), inplace=True)
df["operating_hours_per_day"].fillna(df["operating_hours_per_day"].median(), inplace=True)
df["digital_payment_usage_percent"].fillna(df["digital_payment_usage_percent"].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["business_type"].value_counts()
```

point 2

```python
df.groupby("business_type")["digital_payment_usage_percent"].mean()
```

```python
df["transaction_value_range"] = pd.cut(
    df["transaction_value_inr"],
    bins=[0, 1000, 5000, 10000, float("inf")],
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
df.groupby("transaction_value_range", observed=False)["digital_payment_usage_percent"].mean()
```

point 3

```python
adoption_rate = df["digital_payment_usage_percent"].mean()

print("Digital Payment Adoption Rate:", adoption_rate, "%")
```

```python
df["transaction_volume_group"] = pd.cut(
    df["transaction_count"],
    bins=[0, 100, 300, float("inf")],
    labels=["Low", "Medium", "High"]
)
```

```python
df["transaction_volume_group"].value_counts()
```
